TOSS WIN VS MATCH WIN

```python
# Create a new column:
# Did toss winner also win the match?

matches['toss_match_win'] = np.where(
    matches['toss_winner'] == matches['match_winner'],
    'Yes',
    'No'
)


# Count plot
plt.figure(figsize=(8,5))

sns.countplot(
    data=matches,
    x='toss_match_win'
)

plt.title("Does Winning Toss Help Win the Match?")
plt.xlabel("Toss Winner Also Won Match")
plt.ylabel("Number of Matches")

plt.show()

# Percentage check
toss_impact = matches['toss_match_win'].value_counts(normalize=True) * 100

print("Toss Impact Percentage:")
print(toss_impact)
```

## Analysis: Bat First vs Field First Decision

This visualization shows the toss decision preference of IPL teams after winning the toss.

### Key Insight

The graph clearly shows that most teams prefer to choose **field first** rather than **bat first** after winning the toss.

- Field First: 764 matches
- Bat First: 405 matches

This indicates that captains generally prefer chasing in IPL matches.

### Cricket Understanding

There are several reasons behind this trend:

- Dew factor in night matches makes bowling difficult in the second innings
- Captains get better clarity of the target while chasing
- Flat pitches at venues like Wankhede and Chinnaswamy favor chasing
- Strong batting depth gives teams more confidence while chasing totals

Because of these factors, teams often believe that bowling first provides a tactical advantage.

### Business Understanding

Toss decision is an important predictive feature for match outcome models.

Instead of using only toss winner, combining:

- toss winner
- toss decision
- venue conditions
- chasing history

can significantly improve match prediction accuracy.

### Conclusion

The IPL strongly favors chasing conditions, and this is reflected in the high number of teams choosing to field first after winning the toss.

VENUE WIN PERCENTAGE

Step 1: Total matches per venue

```python
venue_matches = matches['venue'].value_counts().reset_index()
venue_matches.columns = ['venue', 'total_matches']
```

Step 2: Wins per team per venue

```python
venue_team_wins = matches.groupby(['venue', 'winner_name']).size().reset_index(name='wins')
```

Step 3: Merge both

```python
venue_stats = pd.merge(venue_team_wins, venue_matches, on='venue')
```

Step 4: Calculate win percentage

```python
venue_stats['win_percentage'] = (venue_stats['wins'] / venue_stats['total_matches']) * 100
```

Step 5: Get top 10 venues

```python
top_venues = matches['venue'].value_counts().head(10).index
venue_stats = venue_stats[venue_stats['venue'].isin(top_venues)]
```

Step 6: Clean venue names

```python
venue_stats['venue_short'] = venue_stats['venue'].str.split(',').str[0]
```

Step 7: Plot

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=venue_stats,
    x='venue_short',
    y='win_percentage',
    hue='winner_name'
)

plt.title("Team Win Percentage at Top IPL Venues")
plt.xlabel("Venue")
plt.ylabel("Win Percentage")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    r"D:\Projects\ipl-data-science-project\reports\visuals\venue_win_percentage.png",
    bbox_inches="tight"
)

plt.show()

# Preview
print(venue_stats.head(10))
```
