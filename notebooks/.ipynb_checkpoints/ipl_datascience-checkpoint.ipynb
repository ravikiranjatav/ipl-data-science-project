TOSS WIN VS MATCH WIN

```python
# Create a new column:
# Did toss winner also win the match?

matches['toss_match_win'] = np.where(
    matches['toss_winner'] == matches['match_winner'],
    'Yes',
    'No'
)


# Count plot
plt.figure(figsize=(8,5))

sns.countplot(
    data=matches,
    x='toss_match_win'
)

plt.title("Does Winning Toss Help Win the Match?")
plt.xlabel("Toss Winner Also Won Match")
plt.ylabel("Number of Matches")

plt.show()

# Percentage check
toss_impact = matches['toss_match_win'].value_counts(normalize=True) * 100

print("Toss Impact Percentage:")
print(toss_impact)
```

Venue-wise Match Count

```python
# =========================================
#  TOP IPL VENUES
# =========================================

plt.figure(figsize=(12,6))

matches['venue'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 IPL Venues by Number of Matches")
plt.xlabel("Venue")
plt.ylabel("Total Matches")

plt.xticks(rotation=45)

plt.show()

print(matches['venue'].value_counts().head(10))
```

## Analysis: Top IPL Venues by Number of Matches

This visualization shows the top 10 IPL stadiums based on the total number of matches hosted over the years.

### What This Represents

Some stadiums host significantly more IPL matches because they serve as permanent home grounds for major franchises and are also frequently selected for playoff and knockout matches.

The number of matches played at a venue depends on:

- Home ground allocation
- Franchise popularity
- Stadium capacity
- Pitch quality and conditions
- Playoff and final match scheduling
- Neutral venue seasons

This makes certain venues much more important in IPL history.

### Key Insight

The graph identifies the most frequently used IPL venues such as:

- Wankhede Stadium
- Eden Gardens
- M. Chinnaswamy Stadium
- MA Chidambaram Stadium
- Arun Jaitley Stadium

These venues consistently appear at the top because they belong to major IPL franchises and have hosted matches across multiple seasons.

### Cricket Understanding

Venue plays a major role in match outcomes because every stadium behaves differently.

For example:

- Wankhede often supports chasing due to dew
- Chepauk supports spin bowling
- Chinnaswamy is known for high-scoring matches
- Eden Gardens often offers balanced conditions

Because of these variations, teams often perform differently depending on the venue.

### Business Understanding

Venue should be considered a high-impact feature in machine learning models because team performance is strongly influenced by ground conditions.

Using venue as a feature helps improve:

- Match winner prediction
- Toss decision analysis
- Batting vs bowling strategy prediction
- Player performance forecasting

### Conclusion

Venue is not just a location—it directly infl
uences match strategy, scoring patterns, and team success. This makes venue analysis one of the most important parts of IPL data science projects.

TOSS DECISION ANALYSIS

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=matches,
    x='toss_decision'
)

plt.title("Bat First vs Field First Decision")
plt.xlabel("Toss Decision")
plt.ylabel("Number of Matches")

plt.savefig("reports/visuals/toss_decision_analysis.png", bbox_inches="tight")

plt.show()

print("Toss Decision Counts:")
print(matches['toss_decision'].value_counts())
```

## Analysis: Bat First vs Field First Decision

This visualization shows the toss decision preference of IPL teams after winning the toss.

### Key Insight

The graph clearly shows that most teams prefer to choose **field first** rather than **bat first** after winning the toss.

- Field First: 764 matches
- Bat First: 405 matches

This indicates that captains generally prefer chasing in IPL matches.

### Cricket Understanding

There are several reasons behind this trend:

- Dew factor in night matches makes bowling difficult in the second innings
- Captains get better clarity of the target while chasing
- Flat pitches at venues like Wankhede and Chinnaswamy favor chasing
- Strong batting depth gives teams more confidence while chasing totals

Because of these factors, teams often believe that bowling first provides a tactical advantage.

### Business Understanding

Toss decision is an important predictive feature for match outcome models.

Instead of using only toss winner, combining:

- toss winner
- toss decision
- venue conditions
- chasing history

can significantly improve match prediction accuracy.

### Conclusion

The IPL strongly favors chasing conditions, and this is reflected in the high number of teams choosing to field first after winning the toss.
